# World Development Indicators

Load the project-local Excel workbook and take a first look at the available data.

```python
from pathlib import Path

import pandas as pd
```

```python
workbook_path = Path("data/raw/P_WDI_40C_5I_25T.xlsx")

if not workbook_path.exists():
    raise FileNotFoundError("Could not find workbook")

sheets = pd.read_excel(workbook_path, sheet_name=None)
data = sheets["Data"]
metadata = sheets["Series - Metadata"]
```

```python
sheet_summary = pd.DataFrame(
    [
        {"sheet": name, "data_rows_excluding_header": frame.shape[0], "columns": frame.shape[1]}
        for name, frame in sheets.items()
    ]
)

sheet_summary
```

```python
print("Sheet names:", list(sheets))
print("Data shape, excluding header row:", data.shape)
print("Metadata shape, excluding header row:", metadata.shape)
print("Data columns:")
print(data.columns.to_list())
print("Metadata columns:")
print(metadata.columns.to_list())
```
